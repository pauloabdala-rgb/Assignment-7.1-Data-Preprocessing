```python
import pandas as pd
from pathlib import Path

downloads = Path("/Users/macminihique/Downloads")

files = list(downloads.glob("*anada*.xls*"))
print(files)
```

```python
df = pd.read_excel(
    files[0],
    sheet_name="Canada by Citizenship",
    skiprows=20
)

df.head()
```

```python
# Rename columns
df.rename(columns={
    "OdName": "Country",
    "AreaName": "Continent",
    "RegName": "Region"
}, inplace=True)

# Select required columns
years = list(range(1980, 2014))
columns = ["Country", "Continent", "Region", "DevName"] + years

df = df[columns].copy()

# Calculate total immigration
df[years] = df[years].apply(
    pd.to_numeric, errors="coerce"
).fillna(0).astype(int)

df["Total"] = df[years].sum(axis=1)

# Set Country as index
df.set_index("Country", inplace=True)
df.index.name = "Country"

# Display final table
display(df.head())
```
